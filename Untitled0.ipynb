```python
import pandas as pd, numpy as np
pd.set_option("display.max_columns", 60)

df = pd.read_csv("diabetic_data.csv")
print(df.shape)
print(df["readmitted"].value_counts(normalize=True).round(3))
```

```python
df = df.replace("?", np.nan)
df["target"] = (df["readmitted"] == "<30").astype(int)

df = df[~df["discharge_disposition_id"].isin([11, 13, 14, 19, 20, 21])]
df = df.sort_values("encounter_id").drop_duplicates("patient_nbr", keep="first")
df = df[df["gender"] != "Unknown/Invalid"]

df = df.drop(columns=["weight", "payer_code", "medical_specialty",
                      "encounter_id", "patient_nbr", "readmitted"])

print(df.shape, "| positive rate:", df["target"].mean().round(3))
```

```python
import matplotlib.pyplot as plt, seaborn as sns
from scipy import stats

fig, ax = plt.subplots(2, 2, figsize=(12, 9))

df["target"].value_counts().sort_index().plot(kind="bar", ax=ax[0,0])
ax[0,0].set_title("Readmitted within 30 days (0 = no, 1 = yes)")

df.groupby("age")["target"].mean().plot(kind="bar", ax=ax[0,1])
ax[0,1].set_title("Readmission rate by age group")

sns.boxplot(data=df, x="target", y="time_in_hospital", ax=ax[1,0])
ax[1,0].set_title("Days in hospital by outcome")

df.groupby(df["number_inpatient"].clip(upper=5))["target"].mean().plot(kind="bar", ax=ax[1,1])
ax[1,1].set_title("Rate by prior inpatient visits (5 = 5 or more)")

plt.tight_layout(); plt.show()

# Chi-square: age group vs outcome
table = pd.crosstab(df["age"], df["target"])
chi2, p, dof, _ = stats.chi2_contingency(table)
v = np.sqrt(chi2 / (table.values.sum() * (min(table.shape) - 1)))
print(f"Age vs outcome: chi2={chi2:.1f}, p={p:.2e}, Cramer's V={v:.3f}")

# Mann-Whitney: prior inpatient visits, returned vs not
a = df.loc[df.target == 1, "number_inpatient"]
b = df.loc[df.target == 0, "number_inpatient"]
u, p = stats.mannwhitneyu(a, b, alternative="two-sided")
print(f"Prior inpatient: U={u:.0f}, p={p:.2e}, rank-biserial={2*u/(len(a)*len(b))-1:.3f}")
```

```python
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.impute import SimpleImputer

# group ICD-9 diagnosis codes into broad categories
def diag_group(code):
    if pd.isna(code): return "Unknown"
    if str(code).startswith(("V", "E")): return "Other"
    c = float(code)
    if 250 <= c < 251: return "Diabetes"
    if 390 <= c <= 459 or c == 785: return "Circulatory"
    if 460 <= c <= 519 or c == 786: return "Respiratory"
    if 520 <= c <= 579 or c == 787: return "Digestive"
    if 580 <= c <= 629 or c == 788: return "Genitourinary"
    if 710 <= c <= 739: return "Musculoskeletal"
    if 800 <= c <= 999: return "Injury"
    if 140 <= c <= 239: return "Neoplasms"
    return "Other"

df["diag_group"] = df["diag_1"].apply(diag_group)
df["age_num"] = df["age"].str.extract(r"\[(\d+)-")[0].astype(int) + 5   # band midpoint

# 21 medication columns are almost always "No": keep only metformin and insulin
other_drugs = ["repaglinide","nateglinide","chlorpropamide","glimepiride","acetohexamide",
               "glipizide","glyburide","tolbutamide","pioglitazone","rosiglitazone","acarbose",
               "miglitol","troglitazone","tolazamide","examide","citoglipton",
               "glyburide-metformin","glipizide-metformin","glimepiride-pioglitazone",
               "metformin-rosiglitazone","metformin-pioglitazone"]

model_df = df.drop(columns=["diag_1","diag_2","diag_3","age","race"] + other_drugs, errors="ignore")
X = model_df.drop(columns="target").copy()
y = model_df["target"]

# these are numeric codes but really categories
for c in ["admission_type_id", "discharge_disposition_id", "admission_source_id"]:
    X[c] = X[c].astype(str)

num_cols = ["age_num","time_in_hospital","num_lab_procedures","num_procedures","num_medications",
            "number_outpatient","number_emergency","number_inpatient","number_diagnoses"]
cat_cols = [c for c in X.columns if c not in num_cols]

preprocess = ColumnTransformer([
    ("num", StandardScaler(), num_cols),
    ("cat", Pipeline([("imp", SimpleImputer(strategy="constant", fill_value="Missing")),
                      ("ohe", OneHotEncoder(drop="first", handle_unknown="ignore", sparse_output=False))]),
     cat_cols),
])

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42)

print(X.shape, "| train/test positive rate:", y_train.mean().round(3), y_test.mean().round(3))
```

```python
from sklearn.dummy import DummyClassifier
from sklearn.linear_model import LinearRegression, LogisticRegression
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
from sklearn.metrics import roc_auc_score, average_precision_score, brier_score_loss

def fit_pipe(model):
    return Pipeline([("prep", preprocess), ("model", model)]).fit(X_train, y_train)

rows, probs = [], {}

# 1) baseline: everyone gets the training prevalence
base = fit_pipe(DummyClassifier(strategy="prior"))
probs["Baseline (prevalence)"] = base.predict_proba(X_test)[:, 1]

# 2) least squares on 0/1 labels (plug-in classifier)
lin = fit_pipe(LinearRegression())
raw = lin.predict(X_test)
print(f"Least squares outputs: min={raw.min():.3f}, max={raw.max():.3f}, "
      f"share outside [0,1]={((raw < 0) | (raw > 1)).mean():.3f}")
probs["Least squares (clipped)"] = np.clip(raw, 0, 1)

# 3) logistic regression (maximum likelihood, L2 penalty)
logit = fit_pipe(LogisticRegression(max_iter=1000))
probs["Logistic regression"] = logit.predict_proba(X_test)[:, 1]

# 4) LDA (Gaussian classes, shared covariance, Bayes rule)
lda = fit_pipe(LinearDiscriminantAnalysis())
probs["LDA"] = lda.predict_proba(X_test)[:, 1]

for name, p in probs.items():
    rows.append({"Model": name,
                 "ROC-AUC": roc_auc_score(y_test, p),
                 "PR-AUC": average_precision_score(y_test, p),
                 "Brier": brier_score_loss(y_test, p)})

results = pd.DataFrame(rows).round(4)
results
```

```python
from sklearn.base import clone
from sklearn.model_selection import cross_validate
import matplotlib.pyplot as plt

prep = clone(preprocess).fit(X_train)
Xtr = prep.transform(X_train)

Cs = np.logspace(-4, 1, 8)          # C = 1/penalty strength: small C = strong penalty
out = []
for pen in ["l2", "l1"]:
    for C in Cs:
        m = LogisticRegression(penalty=pen, C=C, solver="liblinear", max_iter=1000)
        cv = cross_validate(m, Xtr, y_train, cv=3, scoring="roc_auc", return_train_score=True)
        nz = (m.fit(Xtr, y_train).coef_ != 0).sum()
        out.append({"penalty": pen, "C": C,
                    "train_auc": cv["train_score"].mean(),
                    "val_auc": cv["test_score"].mean(),
                    "nonzero_coefs": nz})
sweep = pd.DataFrame(out).round(4)
print(sweep)

fig, ax = plt.subplots(1, 2, figsize=(12, 4))
for i, pen in enumerate(["l2", "l1"]):
    s = sweep[sweep.penalty == pen]
    ax[i].semilogx(s.C, s.train_auc, "o-", label="train")
    ax[i].semilogx(s.C, s.val_auc, "o-", label="validation")
    ax[i].set_title(f"{pen.upper()} penalty: AUC vs C")
    ax[i].set_xlabel("C (small = stronger penalty)"); ax[i].set_ylabel("ROC-AUC"); ax[i].legend()
plt.tight_layout(); plt.show()
```

```python
from sklearn.neighbors import KNeighborsClassifier

ytr_np = y_train.to_numpy()
rng = np.random.RandomState(42)
idx = rng.permutation(len(Xtr))
tr_i, va_i = idx[:15000], idx[15000:20000]

knn_rows = []
for k in [5, 15, 50, 100, 200, 400]:
    knn = KNeighborsClassifier(n_neighbors=k).fit(Xtr[tr_i], ytr_np[tr_i])
    knn_rows.append({"k": k,
        "train_auc": roc_auc_score(ytr_np[tr_i], knn.predict_proba(Xtr[tr_i])[:, 1]),
        "val_auc":   roc_auc_score(ytr_np[va_i], knn.predict_proba(Xtr[va_i])[:, 1])})
knn_res = pd.DataFrame(knn_rows).round(4)
print(knn_res)

plt.semilogx(knn_res.k, knn_res.train_auc, "o-", label="train")
plt.semilogx(knn_res.k, knn_res.val_auc, "o-", label="validation")
plt.xlabel("k (number of neighbours)"); plt.ylabel("ROC-AUC"); plt.legend()
plt.title("k-NN: bias-variance dial"); plt.show()
```

```python
from sklearn.mixture import GaussianMixture

Xnum = Xtr[:, :len(num_cols)]     # the 9 standardized numeric columns come first

bic = {}
for k in range(1, 8):
    g = GaussianMixture(n_components=k, covariance_type="full", random_state=42, n_init=2).fit(Xnum)
    bic[k] = g.bic(Xnum)
print(pd.Series(bic).round(0))
```

```python
K = 4
gmm = GaussianMixture(n_components=K, covariance_type="full", random_state=42, n_init=3).fit(Xnum)

prof = pd.DataFrame(Xnum, columns=num_cols)
prof["cluster"] = gmm.predict(Xnum)
prof["readmitted"] = y_train.to_numpy()

print(prof.groupby("cluster").agg(size=("readmitted", "size"),
                                  readmit_rate=("readmitted", "mean")).round(3))
print(prof.groupby("cluster")[num_cols].mean().round(2))
```

```python
from sklearn.metrics import roc_curve
from sklearn.calibration import calibration_curve

p  = probs["Logistic regression"]      # test-set probabilities from Cell B
yt = y_test.to_numpy()

# --- ROC curve and calibration curve ---
fig, ax = plt.subplots(1, 2, figsize=(11, 4.5))
fpr, tpr, _ = roc_curve(yt, p)
ax[0].plot(fpr, tpr, label=f"Logistic (AUC = {roc_auc_score(yt, p):.3f})")
ax[0].plot([0, 1], [0, 1], "--", label="Random guessing")
ax[0].set_xlabel("False positive rate"); ax[0].set_ylabel("True positive rate")
ax[0].set_title("ROC curve"); ax[0].legend()

frac_pos, mean_pred = calibration_curve(yt, p, n_bins=10, strategy="quantile")
m = max(mean_pred.max(), frac_pos.max()) * 1.1
ax[1].plot(mean_pred, frac_pos, "o-", label="Logistic")
ax[1].plot([0, m], [0, m], "--", label="Perfect calibration")
ax[1].set_xlabel("Mean predicted probability"); ax[1].set_ylabel("Observed readmission rate")
ax[1].set_title("Calibration curve"); ax[1].legend()
plt.tight_layout(); plt.show()

# --- Bootstrap: resample the test set to see how much the metrics wobble ---
rng = np.random.RandomState(0)
aucs, briers = [], []
for _ in range(1000):
    i = rng.randint(0, len(yt), len(yt))
    aucs.append(roc_auc_score(yt[i], p[i]))
    briers.append(np.mean((p[i] - yt[i]) ** 2))
print(f"AUC   = {roc_auc_score(yt, p):.4f}, 95% bootstrap CI [{np.percentile(aucs, 2.5):.4f}, {np.percentile(aucs, 97.5):.4f}]")
print(f"Brier = {brier_score_loss(yt, p):.4f}, 95% bootstrap CI [{np.percentile(briers, 2.5):.4f}, {np.percentile(briers, 97.5):.4f}]")

# --- Hoeffding bound for the test-set Brier score (squared error lies in [0, 1]) ---
n = len(yt)
eps = np.sqrt(np.log(2 / 0.05) / (2 * n))
print(f"Hoeffding: with 95% confidence, the true Brier score is within +/- {eps:.4f} of the test value")
```

```python
# project-defined risk bands (not medical thresholds)
bands = pd.cut(p, [0, 0.30, 0.60, 1.0], labels=["Low", "Moderate", "High"], include_lowest=True)
print(pd.DataFrame({"band": bands, "y": yt}).groupby("band", observed=False)["y"]
        .agg(patients="size", observed_rate="mean").round(3))
print("max predicted probability:", p.max().round(3))

# do summed probabilities match actual counts in random cohorts of 100?
rng = np.random.RandomState(1)
n_c, reps, exceed, errs = 100, 2000, 0, []
for _ in range(reps):
    i = rng.choice(len(yt), n_c, replace=False)
    expected = p[i].sum()
    sd = np.sqrt((p[i] * (1 - p[i])).sum())
    errs.append(yt[i].sum() - expected)
    exceed += yt[i].sum() > expected + 1.28 * sd
print(f"mean(actual - expected) = {np.mean(errs):.3f} patients per cohort of {n_c}")
print(f"actual exceeded E + 1.28*SD in {exceed/reps:.1%} of cohorts (target: roughly 10%)")
```
